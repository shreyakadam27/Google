```python
import numpy as np
```

```python
x1=np.array([1,1,-1,-1])
y1=np.array([1,-1,1])

x2=np.array([-1,-1,1,1])
y2=np.array([-1,1,-1])
```

```python
print("X1 :",x1)
print("Y1 :",y1)

print("X2 :",x2)
print("Y2 :",y2)
```

```python
w1=np.outer(x1,y1)
w2=np.outer(x2,y2)
w=w1+w2
```

```python
def activation(x):
    return np.where(x>=0,1,-1)
```

```python
def forward(x):
    return activation(np.dot(x,w))
```

```python
def backward(y):
    return activation(np.dot(y,w.T))
```

```python
print("\nFor x1:")
result_y1=forward(x1)
print("y recalled:",result_y1)
result_x1=backward(result_y1)
print("x recalled:",result_x1)

print("\nFor x2:")
result_y2=forward(x2)
print("y recalled:",result_y2)
result_x2=backward(result_y2)
print("x recalled:",result_x2)
```
