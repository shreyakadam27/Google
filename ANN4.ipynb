```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])
```

```python
w = np.array([0.2, -0.1])
b = 0.3
lr = 0.1
epochs = 20

print("Initial Weights:", w)
print("Initial Bias:", b)
```

```python
def activation(net):
    return 1 if net >= 0 else 0
```

```python
for epoch in range(epochs):
    total_error = 0

    for i in range(len(X)):
        output = activation(np.dot(X[i], w) + b)
        error = y[i] - output

        w = w + lr * error * X[i]
        b = b + lr * error

        total_error += abs(error)

    print("Epoch", epoch + 1, "Total Error:", total_error)

    if total_error == 0:
        print("Training stopped after", epoch + 1, "epochs")
        break
```

```python
print("\nFinal Weights:", w)
print("Final Bias:", b)

print("\nVerification")
print("Input\tTarget\tPredicted")

for i in range(len(X)):
    pred = activation(np.dot(X[i], w) + b)
    print(X[i], "\t", y[i], "\t", pred)
```

```python
xx, yy = np.meshgrid(
    np.arange(-0.5, 1.5, 0.01),
    np.arange(-0.5, 1.5, 0.01)
)

Z = (w[0] * xx + w[1] * yy + b >= 0).astype(int)

plt.contourf(xx, yy, Z, alpha=0.3, cmap="coolwarm")

plt.scatter(X[y == 0, 0], X[y == 0, 1],
            color="red", s=100, label="Class 0")

plt.scatter(X[y == 1, 0], X[y == 1, 1],
            color="blue", marker="^", s=100, label="Class 1")

plt.title("Decision Regions Using Perceptron")
plt.xlabel("Input X1")
plt.ylabel("Input X2")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
